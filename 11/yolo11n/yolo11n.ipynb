```python
from ultralytics import YOLO
```

```python
yolov11 = YOLO("yolo11n.pt")
```

```python
metrics = yolov11.train(
    data="Helmet-and-Number-Plate-Detection-for-Motorbike-Safety-1/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    #device=0,                 # GPU
    name="YOLOv11n"
)
```

```python
from ultralytics import YOLO

yolov11_test = YOLO("runs/detect/YOLOv11n/weights/best.pt")#/content/runs/detect/YOLOv11n/weights/best.pt
```

```python
metrics_v11 = yolov11_test.val(
    data="Helmet-and-Number-Plate-Detection-for-Motorbike-Safety-1/data.yaml",
    split="test"
)
```

```python
print("Evaluation Report (YOLOv11n–Test Set)")
print(f"mAP@0.5        : {metrics_v11.box.map50:.3f}")
print(f"mAP@0.5:0.95   : {metrics_v11.box.map:.3f}")
print(f"Precision(mean): {metrics_v11.box.p.mean():.3f}")
print(f"Recall(mean)   : {metrics_v11.box.r.mean():.3f}")
print(f"Inference Speed: {metrics_v11.speed['inference']:.2f} ms/image")
```
