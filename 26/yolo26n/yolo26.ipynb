```python
from ultralytics import YOLO
```

```python
yolov26 = YOLO("yolo26n.pt")
```

```python
metrics = yolov26.train(
    data="Helmet-and-Number-Plate-Detection-for-Motorbike-Safety-1/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    name="YOLOv26n"
)
```

```python
from ultralytics import YOLO

yolov26_test = YOLO("/kaggle/working/runs/detect/YOLOv26n/weights/best.pt")
```

```python
metrics_v26 = yolov26_test.val(
    data="Helmet-and-Number-Plate-Detection-for-Motorbike-Safety-1/data.yaml",
    split="test"
)
```

```python
print("Evaluation Report (YOLOv26n–Test Set)")
print(f"mAP@0.5        : {metrics_v26.box.map50:.3f}")
print(f"mAP@0.5:0.95   : {metrics_v26.box.map:.3f}")
print(f"Precision(mean): {metrics_v26.box.p.mean():.3f}")
print(f"Recall(mean)   : {metrics_v26.box.r.mean():.3f}")
print(f"Inference Speed: {metrics_v26.speed['inference']:.2f} ms/image")
```
